### Data Description

- Customer - Customer ID

- ST - State where customers live

- Gender - Gender of the customer

- Education - Background education of customers 

- Customer Lifetime Value - Customer lifetime value(CLV) is the total revenue the client will derive from their entire relationship with a customer. In other words, is the predicted or calculated value of a customer over their entire duration as a policyholder with the insurance company. It is an estimation of the net profit that the insurance company expects to generate from a customer throughout their relationship with the company. Customer Lifetime Value takes into account factors such as the duration of the customer's policy, premium payments, claim history, renewal likelihood, and potential additional services or products the customer may purchase. It helps insurers assess the long-term profitability and value associated with retaining a particular customer.

- Income - Customers income

- Monthly Premium Auto - Amount of money the customer pays on a monthly basis as a premium for their auto insurance coverage. It represents the recurring cost that the insured person must pay to maintain their insurance policy and receive coverage for potential damages, accidents, or other covered events related to their vehicle.

- Number of Open Complaints - Number of complaints the customer opened

- Policy Type - There are three type of policies in car insurance (Corporate Auto, Personal Auto, and Special Auto)

- Vehicle Class - Type of vehicle classes that customers have Two-Door Car, Four-Door Car SUV, Luxury SUV, Sports Car, and Luxury Car

- Total Claim Amount - the sum of all claims made by the customer. It represents the total monetary value of all approved claims for incidents such as accidents, theft, vandalism, or other covered events.

### Exercise 3

The sales team wants to know if customers with Personal Auto have a lower income than those with Corporate Auto. How does the average income compare between the two policy types?

- Use *loc* to create two dataframes: one containing only Personal Auto policies and one containing only Corporate Auto policies.
- Calculate the average income for each policy.
- Print the results.

```python
import pandas as pd

# Load the raw dataset
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)

# Clean column headers for consistency (removes extra spaces)
df.columns = df.columns.str.strip()

# ==============================================================================
# STEP 1: Filter data using .loc to create two distinct DataFrames
# ==============================================================================

# WHAT: Create a DataFrame containing only customers with 'Personal Auto' policies.
# WHY: Using .loc allows us to filter rows where the 'Policy Type' column matches 'Personal Auto'.
personal_auto_df = df.loc[df['Policy Type'] == 'Personal Auto']

# WHAT: Create a DataFrame containing only customers with 'Corporate Auto' policies.
# WHY: Isolating 'Corporate Auto' records into a separate DataFrame allows independent calculations.
corporate_auto_df = df.loc[df['Policy Type'] == 'Corporate Auto']


# ==============================================================================
# STEP 2: Calculate the average income for each policy type
# ==============================================================================

# WHAT: Compute the arithmetic mean of the 'Income' column for Personal Auto customers.
# WHY: Provides the baseline average earning figure for Personal Auto policyholders.
personal_avg_income = personal_auto_df['Income'].mean()

# WHAT: Compute the arithmetic mean of the 'Income' column for Corporate Auto customers.
# WHY: Provides the baseline average earning figure for Corporate Auto policyholders.
corporate_avg_income = corporate_auto_df['Income'].mean()


# ==============================================================================
# STEP 3: Print the results and comparison
# ==============================================================================

print("--- AVERAGE INCOME BY POLICY TYPE ---")
print(f"Personal Auto Average Income:  ${personal_avg_income:,.2f}")
print(f"Corporate Auto Average Income: ${corporate_avg_income:,.2f}")

# Compare and print key finding for the sales team
difference = corporate_avg_income - personal_avg_income
print(f"\nConclusion: Corporate Auto policyholders have a higher average income by ${difference:,.2f}.")
```

```python
import pandas as pd

# Load the raw dataset
url = "https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv"
df = pd.read_csv(url)

# Clean column headers for consistency (removes trailing or leading spaces)
df.columns = df.columns.str.strip()


# ==============================================================================
# STEP 1: Calculate the 75th Percentile (0.75 Quantile)
# ==============================================================================

# WHAT: Use the .quantile(0.75) method on the 'Total Claim Amount' column.
# WHY: The 75th percentile marks the threshold above which the top 25% highest claim amounts fall.
threshold_75th = df['Total Claim Amount'].quantile(0.75)

print("--- STEP 1: PERCENTILE THRESHOLD ---")
print(f"75th Percentile Threshold for Total Claim Amount: ${threshold_75th:,.2f}")


# ==============================================================================
# STEP 2: Filter the DataFrame using Boolean Selection (.loc)
# ==============================================================================

# WHAT: Filter the original DataFrame to include only rows where 'Total Claim Amount' > threshold_75th.
# WHY: This creates a new DataFrame isolating high-value claim customers (the top 25%) 
# for profitability and risk analysis.
high_claims_df = df.loc[df['Total Claim Amount'] > threshold_75th]


# ==============================================================================
# STEP 3: Calculate Summary Statistics for High Claim Customers
# ==============================================================================

# WHAT: Use .describe() on the filtered DataFrame to calculate key summary metrics.
# WHY: Analyzing count, mean, std, min, median (50%), and max helps understand the distribution 
# and total financial impact of these high-claim policyholders.
print("\n--- STEP 3: SUMMARY STATISTICS FOR HIGH CLAIM CUSTOMERS (TOP 25%) ---")
high_claims_stats = high_claims_df['Total Claim Amount'].describe()

# Add median explicitly for central tendency comparison
high_claims_median = high_claims_df['Total Claim Amount'].median()

print(high_claims_stats)
print(f"Median Claim Amount: ${high_claims_median:,.2f}")


# Optional: Display the first few rows of high claim customers
print("\n--- SAMPLE OF HIGH CLAIM CUSTOMERS ---")
display(high_claims_df[['Customer', 'ST', 'Policy Type', 'Income', 'Total Claim Amount']].head())
```
